**Cleaning Data**

Importing Dependencies

```python
import pandas as pd
```

*Lendeborg Data*

```python
# Import Data from CSV File
lendeborg = pd.read_csv("/Users/jessicalo/Documents/sports_analysis_project/data/raw/lendeborg_stats.csv")
# Drop Final Column (Totals)
lendeborg = lendeborg.drop(lendeborg.index[-1])
```

```python
# Fix column regarding venue type, renaming it, and changing values to improve readability
lendeborg = lendeborg.rename(columns={"Unnamed: 5": "Venue"})
lendeborg['Venue'] = lendeborg['Venue'].astype(str)
lendeborg["Venue"] = lendeborg["Venue"].fillna("Home")
lendeborg.loc[lendeborg["Venue"] == "@", "Venue"] = "Away"
lendeborg.loc[lendeborg["Venue"] == "N", "Venue"] = "Neutral"
```

```python
# Convert Date column to datetime type
lendeborg["Date"] = pd.to_datetime(lendeborg["Date"])
```

```python
# Add columns Outcome and Player for later joins
lendeborg["Outcome"] = lendeborg["Result"].str[0]
lendeborg["Player"] = "Yaxel Lendeborg"
```
