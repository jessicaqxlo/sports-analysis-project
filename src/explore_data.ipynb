**Cleaning Data**

Importing Dependencies

```python
import pandas as pd
```

*Lendeborg Data*

```python
# Import Data from CSV File
lendeborg = pd.read_csv("/Users/jessicalo/Documents/sports_analysis_project/data/raw/lendeborg_stats.csv")
# Drop Final Column (Totals)
lendeborg = lendeborg.drop(lendeborg.index[-1])
```

```python
# Fix column regarding venue type, renaming it, and changing values to improve readability
lendeborg = lendeborg.rename(columns={"Unnamed: 5": "Venue"})
lendeborg['Venue'] = lendeborg['Venue'].astype(str)
lendeborg["Venue"] = lendeborg["Venue"].fillna("Home")
lendeborg.loc[lendeborg["Venue"] == "@", "Venue"] = "Away"
lendeborg.loc[lendeborg["Venue"] == "N", "Venue"] = "Neutral"
```

```python
# Convert Date column to datetime type
lendeborg["Date"] = pd.to_datetime(lendeborg["Date"])
```

```python
# Add columns Outcome and Player for later joins
lendeborg["Outcome"] = lendeborg["Result"].str[0]
lendeborg["Player"] = "Yaxel Lendeborg"
```

*Johnson Jr. Data*

```python
# Import Data from CSV File
johnson = pd.read_csv("/Users/jessicalo/Documents/sports_analysis_project/data/raw/johnson_stats.csv")
# Drop Final Column (Totals)
johnson = johnson.drop(johnson.index[-1])
```

```python
# Fix column regarding venue type, renaming it, and changing values to improve readability
johnson = johnson.rename(columns={"Unnamed: 5": "Venue"})
johnson['Venue'] = johnson['Venue'].astype(str)
johnson["Venue"] = johnson["Venue"].fillna("Home")
johnson.loc[johnson["Venue"] == "@", "Venue"] = "Away"
johnson.loc[johnson["Venue"] == "N", "Venue"] = "Neutral"
```

```python
# Convert Date column to datetime type
johnson["Date"] = pd.to_datetime(johnson["Date"])
```

```python
# Add columns Outcome and Player for later joins
johnson["Outcome"] = johnson["Result"].str[0]
johnson["Player"] = "Morez Johnson Jr."
```

*Write Data to New CSV File*

```python
stacked_table = pd.concat([lendeborg, johnson], ignore_index=True)
stacked_table.to_csv("/Users/jessicalo/Documents/sports_analysis_project/data/processed/processed_data.csv", index=False)
```
